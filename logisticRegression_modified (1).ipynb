2. Data Preprocessing:

a. Handle missing values (e.g., imputation).

```python
X = train_df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1)
# Impute missing 'Age' values with the mean
train_df['Age'].fillna(train_df['Age'].mean(), inplace=True)

# Impute missing 'Fare' values with the mean
train_df['Fare'].fillna(train_df['Fare'].mean(), inplace=True)

# Handle missing Cabin values (if any) by filling with 'Unknown'
train_df['Cabin'].fillna('Unknown', inplace=True)

train_df['Embarked'].fillna(train_df['Embarked'].mode()[0], inplace=True)

# Encode categorical variables
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1, 'female': 1, 'Male': 0, 'Female': 1})

# One-hot encode 'Embarked' and store the column names
X = pd.get_dummies(X, columns=['Embarked'], prefix='Embarked')

# Verify that there are no more missing values
print("\nMissing values after imputation:")
print(train_df.isnull().sum())
```

```python
print(train_df.columns)
```

b. Encode categorical variables

```python
train_df["Sex"] = train_df["Sex"].astype(str)
test_df["Sex"] = test_df["Sex"].astype(str)

# Encode 'Sex' using LabelEncoder
sex_encoder = LabelEncoder()
combined_sex_data = pd.concat([train_df["Sex"], test_df["Sex"]], axis=0)
sex_encoder.fit(combined_sex_data)

train_df["Sex"] = sex_encoder.transform(train_df["Sex"])
test_df["Sex"] = sex_encoder.transform(test_df["Sex"])
```

```python
# Handle 'Cabin' - create a simplified feature (e.g., deck) or drop it
# Example: Extract the first letter of the cabin
train_df['Cabin'] = train_df['Cabin'].astype(str).str[0]
cabin_dummies = pd.get_dummies(train_df['Cabin'], prefix='Cabin')
train_df = pd.concat([train_df, cabin_dummies], axis=1)
train_df.drop('Cabin', axis=1, inplace=True)
```

```python
train_df = pd.get_dummies(train_df, columns=['Embarked'], drop_first=True, dtype=int)
```

```python
# Remove unnecessary features such as 'Name' and 'Ticket'
train_df.drop(['Name', 'Ticket'], axis=1, inplace=True)
```

```python
print(train_df.head())
```

```python
features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked_Q", "Embarked_S"]
target = 'Survived'
X = train_df[features]
y = train_df[target]
feature_names = X.columns
```

3. Model Building:

a. Build a logistic regression model using appropriate libraries (e.g., scikit-learn).

```python
# Scale the numerical features
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
y_train = train_df['Survived']
```

```python
# Build the model
model = LogisticRegression(random_state=42)
```

```python
model.fit(X,y_train)
```

```python
joblib.dump(model, 'logistic_regression_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve
import matplotlib.pyplot as plt
```

```python
# Make predictions on the same data (since there is no separate test set)
y_pred = model.predict(X) # Use X instead of X_test
y_pred # y hat values for x
```

```python
accuracy = accuracy_score(y, y_pred)
precision = precision_score(y, y_pred, zero_division=0)
recall = recall_score(y, y_pred, zero_division=0)
f1 = f1_score(y, y_pred, zero_division=0)
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1-Score: {f1}")
```

a.Interpret the coefficients of the logistic regression model

Pclass: A negative coefficient suggests that as the passenger class increases (e.g., from 1st to 3rd class), the log-odds of survival decrease. This implies that higher passenger classes (1st class) were more likely to survive.
    
Sex: If Sex is encoded as 0 for male and 1 for female, a positive coefficient suggests that being female increases the log-odds of survival.
    
Age: A negative coefficient suggests that as age increases, the log-odds of survival decrease. This implies that younger passengers were more likely to survive.

SibSp: A negative coefficient suggests that having more siblings/spouses aboard decreases the log-odds of survival.

Parch: A negative coefficient suggests that having more parents/children aboard decreases the log-odds of survival.

Fare: A positive coefficient suggests that as the fare increases, the log-odds of survival increase. This implies that passengers who paid higher fares were more likely to survive.

Embarked_Q: A negative coefficient suggests that embarking from Queenstown decreases the log-odds of survival compared to the reference category (which is Embarked_C because it was dropped).

b. Discuss the significance of features in predicting the target variable (survival probability in this case).

Sex: The 'Sex' feature is a strong predictor of survival. Generally, females had a higher survival rate than males. This could be attributed to the "women and children first" policy during the evacuation. The magnitude and sign of the coefficient for 'Sex' will show the extent to which being female or male influenced survival.

Pclass: 'Pclass' (Passenger Class) is a significant predictor. Passengers in higher classes (1st class) had better survival rates. This is likely because they were located closer to the upper decks and had better access to lifeboats. The coefficient will show how much higher or lower survival was based on passenger class.

Age: 'Age' is a relevant factor, with younger passengers generally having a higher chance of survival. The coefficients suggest that children had a higher survival rate, potentially due to prioritization during evacuation.

SibSp and Parch: 'SibSp' (Number of Siblings/Spouses Aboard) and 'Parch' (Number of Parents/Children Aboard) indicate family size. These features can influence survival, but their impact may depend on specific circumstances (e.g., families staying together).

Fare: 'Fare' is an indicator of economic status. Passengers who paid higher fares typically had a higher likelihood of survival. This is likely because wealthier passengers had access to better accommodations and were prioritized during evacuation.

Embarked: 'Embarked' (Port of Embarkation) is a less direct predictor but can reveal differences in survival rates based on the port of embarkation.

```python
# Train-Test Split and Model Evaluation
from sklearn.metrics import classification_report, accuracy_score

# Feature selection
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
train_df['Sex'] = LabelEncoder().fit_transform(train_df['Sex'])
train_df['Age'].fillna(train_df['Age'].mean(), inplace=True)

X = train_df[features]
y = train_df['Survived']

# Standardization
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Train logistic regression
model = LogisticRegression()
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("Accuracy on test data:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
